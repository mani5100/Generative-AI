```python
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
```

# Step 1 Tool Creation

```python
@tool
def multiply(a:int,b:int)->int:
    "This funtion will multiply 2 numbers"
    return a*b
```

# Step 2 Tool Binding

```python
llm=ChatOpenAI()
binded_llm=llm.bind_tools([multiply])
```

# Step 3 ToolCalling

```python
res=binded_llm.invoke("I want to multiply 2 with 4")
res
```

```python
res.tool_calls
```

```python
res.tool_calls[0]['args']
```

# Step 4 Tool Execution

If i call tool with only args i will get the answer

```python
multiply.invoke(res.tool_calls[0]['args'])
```

Butif we call it giving the full object it will give us the ToolMessage

```python
multiply.invoke(res.tool_calls[0])
```

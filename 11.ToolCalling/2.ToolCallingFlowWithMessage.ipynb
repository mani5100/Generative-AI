```python
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
```

# Tool creation

```python
@tool
def multiply(a:int,b:int)->int:
    "This function will Multiply 2 numbers"
    return a*b
```

# Tool Binding

```python
llm =ChatOpenAI()
binded_llm=llm.bind_tools([multiply])
```

# Tool Calling

```python
query=HumanMessage(content="what is multiplication of 3 and 4")
```

```python
messages=[query]
```

```python
AIMessage=binded_llm.invoke(messages)
messages.append(AIMessage)
```

```python
tool_message=multiply.invoke(AIMessage.tool_calls[0])
tool_message
```

```python
messages.append(tool_message)
messages
```

```python
response=binded_llm.invoke(messages)
response.content
```

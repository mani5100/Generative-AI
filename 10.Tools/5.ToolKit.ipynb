```python
from langchain_core.tools import tool
```

```python
@tool
def multiply(a:int,b:int)->int:
    "Multiply two numbers"
    return a*b
@tool
def add(a:int,b:int)->int:
    "Add two numbers"
    return a+b
```

```python
class MathToolKit:
    def get_math_tools(self):
        return [add,multiply]
```

```python
MathTools=MathToolKit()
tool=MathTools.get_math_tools()
```

```python
tool[1].invoke({
    "a":5,
    "b":5
})
```

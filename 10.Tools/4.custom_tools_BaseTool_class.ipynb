```python
from langchain_core.tools import BaseTool
from pydantic import BaseModel, Field
from typing import Type
```

```python
class MultiplyClass(BaseModel):
    a:int=Field(required=True, description="First Number to Multiply")
    b:int=Field(required=True, description="second Number to Multiply")
```

```python
class MultiplyTool(BaseTool):
    name:str="Multiplty"
    description:str="Multiplty 2 Numbers"
    args_schema:Type[BaseModel]=MultiplyClass
    def _run(self,a:int,b:int)->int:
        return a*b
```

```python
tool=MultiplyTool()
res=tool.invoke({
    "a":5,
    "b":10
})
print(res)
```
